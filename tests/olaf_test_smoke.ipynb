## Protocol after the guard

1. Confirm the target is isolated and non-production without copying identifiers into the repository.
2. Confirm the control store's external-access review. Do not upload a real workbook; use only synthetic principals and data.
3. Recompute the per-run workspace-isolation attestation. This is separate from the technical DAR snapshot plus immutable ETag.

5. Run the sentinel immediately before the first write.
